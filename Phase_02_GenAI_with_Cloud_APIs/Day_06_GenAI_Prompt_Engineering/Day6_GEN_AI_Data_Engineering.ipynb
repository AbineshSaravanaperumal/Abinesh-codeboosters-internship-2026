```python
from transformers import pipeline
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
text = input("Enter text: ")
result = classifier(text)
print(result)
```

```python
from transformers import pipeline
generator = pipeline("text-generation", model="gpt2")
text = "We are one in a society"
result = generator(text, max_new_tokens=80,do_sample=True,temperature=0.8,top_p=0, num_return_sequences=2)

for i,out in enumerate(result):
  print(f"---generate {i}---")
  print(out['generated_text'])
```

```python
from getpass import getpass
import os
HF_TOKEN = getpass("Enter your Hugging Face token: ")
os.environ["HF_TOKEN"] = HF_TOKEN
```

```python
from huggingface_hub import InferenceClient

MODEL_ID = "meta-llama/Meta-Llama-3-8B-Instruct"

client = InferenceClient(api_key= os.environ["HF_TOKEN"],model=MODEL_ID)

def extract_text(resp):
  try:
    return resp.choices[0].message.content
  except Exeption:
    try:
      return resp['choices'][0]['message']['content']
    except:
      return str(resp)

messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "What's the current updates in AI"},
]

resp = client.chat_completion(messages=messages)
print(extract_text(resp))
```
