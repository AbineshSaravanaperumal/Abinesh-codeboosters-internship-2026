```python
import random

random_row_indices = df.sample(n=5, random_state=42).index
column_names = df.columns.tolist()

for row_index in random_row_indices:
    random_column = random.choice(column_names)
    df.loc[row_index, random_column] = 0

print("DataFrame after setting 5 new random fields to 0:")
display(df)
```

```python
df.isnull()
```

```python
print("Null values after the operation:")
df.isnull().sum()
```

```python
import numpy as np

df_math_nans = df.copy()

random_row_indices_math_nan = df_math_nans.sample(n=5, random_state=42).index

for row_index in random_row_indices_math_nan:
    df_math_nans.loc[row_index, 'math_score'] = np.nan

print("DataFrame after setting 5 random fields in 'math_score' to `np.nan`:")
display(df_math_nans)
```
